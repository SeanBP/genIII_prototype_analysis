```python
import pickle
import pandas as pd
import numpy as np
```

```python
data_df = pd.read_pickle("e+5GeV_DF.pkl")
fraction_df = pd.read_pickle("../ZDC_Board_Tests/crosstalk_fractions.pkl")
```

```python
# Convert crosstalk DataFrame to NumPy array for fast access
crosstalk_matrix = fraction_df.to_numpy()  # shape: (25, 25)

# Initialize new column
data_df["energy_MIP_xtalk"] = 0.0

# Function to apply crosstalk redistribution per (event, layer)
def apply_xtalk(group):
    # Create an array of energy values by channel
    energies = np.zeros(25)
    for i, row in group.iterrows():
        ch = row["layer_ch"]
        if 0 <= ch < 25:
            energies[int(ch)] = row["energy_MIP"]

    # Redistribute using matrix multiplication: new_energy = M @ energies
    redistributed = crosstalk_matrix @ energies

    # Assign new values to the dataframe
    group = group.copy()
    group["energy_MIP_xtalk"] = [redistributed[int(ch)] if 0 <= ch < 25 else 0.0 for ch in group["layer_ch"]]
    return group

# Apply the redistribution per (event, layer)
data_df = data_df.groupby(["event", "layer"]).apply(apply_xtalk).reset_index(drop=True)
data_df["energy_MIP"] = data_df["energy_MIP_xtalk"]
print(data_df)
```

```python
with open('e+5GeV_ct_DF.pkl', 'wb') as f:
    pickle.dump(data_df, f)
```
