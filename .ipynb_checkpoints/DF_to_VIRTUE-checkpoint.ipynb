```python
import pickle
import os
import numpy as np
import pandas as pd
import json
```

```python
df = pd.read_pickle("BNLCosmic_DF.pkl")
```

```python
# Make a mask for the two conditions
mask = ((df['layer'] == 0) & (df['layer_ch'] == 6)) | \
       ((df['layer'] == 1) & (df['layer_ch'] == 1))

# Set LG_ped_corr to 0 where the mask is True
df.loc[mask, 'LG_ped_corr'] = 0
df.loc[mask, 'HG_ped_corr'] = 0
```

```python
# Generate the header section
header = {
    "version": "2.2.0",
    "experiment": f"ZDC Prototype",
    "energy_unit": "ADC", # Select "eV", "KeV", "MeV", "GeV", or "TeV"
    "color_bar": "Lin", # Select "Lin" or "Log"
    "length_unit": "cm",
    "time_scale": 10,
    "particles": [
        {
        "angle_rad": [0.0 , 0.0],       
        "size": float(10),
        "color_rgba": [float(1), float(0), float(0), float(1)]
        }
    ],
    "tracker_settings": {
        "B_field_T": float(0),
        "tracker_boundary": [float(1), float(1), float(1)]
    }
}

# Main data structure
data = {
    "header": header,
    "events": []
}
```

```python
num_events_to_include = 100  # Change this as needed

# Group the DataFrame by event number
grouped = df.groupby('event')

events = []
maxE = df['HG'].max()  # or set manually, e.g., maxE = 1000.0
print(maxE)
for i, (evt_num, group) in enumerate(grouped):
    if i >= num_events_to_include:
        break  # Stop once the requested number of events is processed

    info_text = f"Event #{evt_num}"
    event = {
        "event_data": {
            "info_text": info_text,
            "energy_scale": [0, float(maxE)]
        },
        "hits": [],
        "clusters": [],
        "tracks": [],
        "jets": [],
        "blocks": []
    }
    
    blocks = []
    for _, row in group.iterrows():
        energy = row['LG_ped_corr']
        if energy <= 0:
            continue  # Skip zero or negative energies
        
        x, y, z = row['x'], row['y'], row['z']
        normalized_energy = energy / maxE
        normalized_energy = (np.log(energy) - np.log(1)) / (np.log(maxE) - np.log(1))
        red, green, blue, alpha = normalized_energy, 0, 1 - normalized_energy, normalized_energy
        c = 299.792  # speed of light in mm/ns
        time = z / c

        block = {
            "position": [float(x), float(y), float(z)],
            "time_ns": float(time),
            "size": [48.8, 48.8, 4],
            "color_rgba": [float(red), float(green), float(blue), float(alpha)]
        }
        blocks.append(block)
    
    event["blocks"] = blocks
    events.append(event)

# Final output dictionary
data = {
    "header": header,
    "events": events
}
```

```python
eventPath = f'/Users/seanbp/Library/Application Support/Steam/steamapps/common/VIRTUE/VIRTUE_macOS.app/Contents/Resources/Data/StreamingAssets/Events/BNLCosmic2.json'
with open(eventPath, 'w') as outfile:
    json.dump(data, outfile, indent=4)
```
